## Reshapping, stacking, squeezing, and unsqueezing tensors

```python
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
#Reshape, add an extra dimension MUST BE COMPATBLE WITH ORIGINAL SIZE
x_reshape = x.reshape(9,1)
x_reshape, x_reshape.shape
```

```python
#Change the view - shares the memory as the orignial tensor
z = x.view(1,9)
z, z.shape
```

```python
#What if z changes?
z[:, 0] = 5
z,x
```

##

```python
#Stack tensors on top of each other
x_stacked = torch.stack([x,x,x,x], dim=0)
x_stacked
```

```python
#Squeeze - removes all single dimensions
x = torch.zeros(2, 1, 2, 1, 2)
x.size()
torch.Size([2, 1, 2, 1, 2])
y = torch.squeeze(x)
y.size()
torch.Size([2, 2, 2])
y = torch.squeeze(x, 0)
y.size()
torch.Size([2, 1, 2, 1, 2])
y = torch.squeeze(x, 1)
y.size()
torch.Size([2, 2, 1, 2])
y = torch.squeeze(x, (1, 2, 3))
torch.Size([2, 2, 2])
```

```python
x_reshape.squeeze().shape
```

```python
#Unsqueeze - adds a single dimensions at a specific dimensions of a tensor
print(f"Previouse: {x.squeeze()}")
print(f"Shape: {x.squeeze().shape}")

x = x.squeeze()

x_unsqueezed = x.unsqueeze(dim=0)
x_unsqueezed
```

```python
#torch.permute - rearranges dimensions in a specified order
x_original = torch.rand(size=(224,224,3)) #image data standards

#Permute to rearrange the axis order
x_permuted = x_original.permute(2,0,1) #Shifts the axis to a new location

#Will have the same place in memory
print(f"Old: {x_original.shape}")
print(f"New: {x_permuted.shape}")
```

##Pytorch and NumPy

* Data in NumPy -> Pytorch tensor
* tensor -> NumPy

```python
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)

#Same data but differnt data type by default
array, tensor
```

```python
#Change the value of array, tensor difference?
array = array + 1

#Will not change the tensor
array,tensor
```

```python
#Tensor -> NumPy
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
#Change tensor, array change?
tensor = tensor + 1

#Does not change
tensor, numpy_tensor
```

## Reproducibility
* Taking the random out of random

```python
#Random numbers -> operations -> update random numbers and make them better -> again...
torch.rand(3,3)
```

```python
#Random seeds reduce some of the overal randomness by flavoring it
import torch
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A==random_tensor_B)
```

```python
#random but reproducible tensors
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

random_tensor_C = torch.rand(3,4)

#need to reset the seed after each use
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
