##Creating range of tensors and tensors-like

```python
torch.arange(1,11) #will replace range
```

```python
one_to_ten = torch.arange(start=1,end=11, step=1)
one_to_ten
```

## Tensor datatypes
One of the 3 areas where erros occure
1. Wrong datatype
2. Wrong shaope
3. Not of the right device

```python
#Float 32
```

```python
#The parameters of a tensor
float_32_tensor = torch.tensor([3.0,6.0,9.0],
                                dtype=None, #The datatype
                                device=None, #for gpu set to cuda, erros occure if data is not on same type of device
                                requires_grad=False) #Weather or not to track gradients

float_32_tensor.dtype
```

```python
#can convert a tensor to a diffent dtype
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```
